### Q1. What is a database? 

Differentiate between SQL and NoSQL databases. 

A **database** is an organized, electronic collection of structured data that allows for efficient storage, retrieval, and management of information. It is typically controlled by a **Database Management System (DBMS)**.

| Feature | SQL Databases (RDBMS) | NoSQL Databases (Non-Relational) |
| --- | --- | --- |
| **Data Model** | Relational; data is stored in tables with rows and columns. | Non-relational; data can be document, key-value, or graph-based. |
| **Schema** | **Static/Predefined:** The structure must be defined before adding data. | **Dynamic:** Schemas can be changed on the fly for unstructured data. |
| **Scaling** | **Vertical:** Handled by increasing the CPU, RAM, or SSD on a single server. | **Horizontal:** Handled by adding more servers to the database cluster. |
| **Integrity** | Follows **ACID** properties (Atomicity, Consistency, Isolation, Durability). | Follows **BASE** (Basically Available, Soft state, Eventual consistency). |

### Q2. What is DDL? 

Explain why CREATE, DROP, ALTER, and TRUNCATE are used with an example. 

**DDL (Data Definition Language)** consists of SQL commands used to define or modify the structure (schema) of the database.

* 
**CREATE**: Used to build new database objects like tables or indexes.


* 
*Example:* `CREATE TABLE Students (Roll_No INT, Name VARCHAR(20));` 




* 
**DROP**: Used to completely delete a table or database from the system.


* 
*Example:* `DROP TABLE Students;` 




* 
**ALTER**: Used to modify an existing table's structure, such as adding or renaming columns.


* 
*Example:* `ALTER TABLE Students ADD Email VARCHAR(50);` 




* 
**TRUNCATE**: Used to remove all records from a table while keeping the table structure intact.


* 
*Example:* `TRUNCATE TABLE Students;`

### Q3. What is DML? 

Explain INSERT, UPDATE, and DELETE with an example. 

**DML (Data Manipulation Language)** refers to commands used to manage and manipulate the data stored within the tables.

* 
**INSERT**: Used to add new rows of data.


* 
*Example:* `INSERT INTO Students VALUES (1, 'Alice');` 




* 
**UPDATE**: Used to modify existing data in a table.


* 
*Example:* `UPDATE Students SET Name = 'Bob' WHERE Roll_No = 1;` 




* 
**DELETE**: Used to remove specific rows based on a condition.


* 
*Example:* `DELETE FROM Students WHERE Roll_No = 1;`

### Q4. What is DQL? 

Explain SELECT with an example. 

**DQL (Data Query Language)** is used to perform queries on the data to retrieve information.

* 
**SELECT**: This command is used to fetch data from the database. You can retrieve specific columns or use `*` for all columns.


* 
*Example:* `SELECT Name FROM Students WHERE Roll_No = 1;`

### Q5. 

Explain Primary Key and Foreign Key. 

* 
**Primary Key**: A column or a set of columns that uniquely identifies each row in a table. It must contain unique values and cannot contain NULLs.


* 
**Foreign Key**: A field in one table that refers to the Primary Key in another table. It establishes a link between the data in the two tables and ensures referential integrity.

### Q6. Write a python code to connect MySQL to python. 

Explain the cursor() and execute() method. 

To connect Python to MySQL, we use the `mysql-connector-python` library.

```python
import mysql.connector

# Establishing the connection
db_connection = mysql.connector.connect(
  host="localhost",
  user="your_username",
  password="your_password",
  database="school_db"
)

# 1. cursor() method: Creates an object used to execute SQL queries 
# [cite_start]and fetch results from the database[cite: 14].
db_cursor = db_connection.cursor()

# 2. execute() method: Sends the SQL statement to the database 
# [cite_start]to be processed[cite: 14].
db_cursor.execute("SELECT * FROM Students")

# Fetching and printing results
for record in db_cursor.fetchall():
    print(record)

db_connection.close()

```

### Q7. 

Give the order of execution of SQL clauses in an SQL query. 

Although we write a query starting with `SELECT`, the database engine processes the clauses in a specific logical order to handle data filtering and grouping efficiently.

1. 
**FROM / JOIN**: Identifying and combining source tables.


2. 
**WHERE**: Filtering individual rows.


3. 
**GROUP BY**: Grouping rows that have the same values.


4. 
**HAVING**: Filtering groups based on aggregate conditions.


5. 
**SELECT**: Calculating expressions and choosing columns.


6. **DISTINCT**: Removing duplicate results.
7. 
**ORDER BY**: Sorting the final result set.


8. 
**LIMIT / OFFSET**: Restricting the number of rows returned.
