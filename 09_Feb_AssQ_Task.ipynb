### Q1. Create a vehicle class with an init method.

In Object-Oriented Programming, the `__init__` method is a constructor used to initialize the state of an object.

```python
class vehicle:
    def __init__(self, name_of_vehicle, max_speed, average_of_vehicle):
        # [cite_start]Instance variables initialized here [cite: 29]
        self.name_of_vehicle = name_of_vehicle
        self.max_speed = max_speed
        self.average_of_vehicle = average_of_vehicle

# Example of creating an instance
my_vehicle = vehicle("Tesla Model 3", 250, 15)
print(f"Vehicle: {my_vehicle.name_of_vehicle}, Speed: {my_vehicle.max_speed}")

```

### Q2. Create a child class 'car' with a 'seating_capacity' method.

Inheritance allows a child class to access attributes and methods of a parent class.

```python
class car(vehicle):
    # [cite_start]Method to return name and capacity [cite: 31]
    def seating_capacity(self, capacity):
        self.capacity = capacity
        return f"Vehicle Name: {self.name_of_vehicle}, Seating Capacity: {self.capacity}"

# Example usage
my_car = car("Toyota Fortuner", 180, 10)
print(my_car.seating_capacity(7))

```

### Q3. What is multiple inheritance? Write a python code to demonstrate it.

**Multiple Inheritance** is a feature where a single class can inherit attributes and methods from more than one parent class.

```python
# Parent Class 1
class Engine:
    def engine_info(self):
        return "This is a V8 engine."

# Parent Class 2
class Body:
    def body_info(self):
        return "This is a Sedan body type."

# [cite_start]Child Class inheriting from both Engine and Body [cite: 32]
class LuxuryCar(Engine, Body):
    def car_info(self):
        return "This is a Luxury Car with combined features."

# Demonstration
obj = LuxuryCar()
print(obj.engine_info())
print(obj.body_info())
print(obj.car_info())

```

### Q4. What are getter and setter in python? Create a class with both.

**Getters** and **Setters** are methods used to access and modify private variables in a class, ensuring data encapsulation. They allow for validation before updating a value.

```python
class Student:
    def __init__(self, age):
        self.__age = age  # Private variable

    # [cite_start]Getter method to retrieve age [cite: 33]
    @property
    def age(self):
        return self.__age

    # [cite_start]Setter method to modify age with validation [cite: 33]
    @age.setter
    def age(self, value):
        if value > 0:
            self.__age = value
        else:
            print("Age must be positive!")

# Demonstration
s = Student(20)
s.age = 25  # Uses setter
print(f"Student Age: {s.age}") # Uses getter

```

### Q5. What is method overriding in python? Write a python code to demonstrate it.

**Method Overriding** occurs when a child class provides a specific implementation for a method that is already defined in its parent class. This allows the child class to modify the behavior of the parent method.

```python
class Animal:
    def make_sound(self):
        return "Generic animal sound"

class Dog(Animal):
    # [cite_start]Overriding the parent method [cite: 34]
    def make_sound(self):
        return "Bark"

class Cat(Animal):
    # [cite_start]Overriding the parent method [cite: 34]
    def make_sound(self):
        return "Meow"

# Demonstration
animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.make_sound())

```
